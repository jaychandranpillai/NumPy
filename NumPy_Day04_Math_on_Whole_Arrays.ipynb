# NumPy Day 4: Math on Whole Arrays

**Topics:** array math with a number, array math with another array, summary functions (`sum`, `mean`, `max`, `min`), and `axis` on 2D arrays.

## Subtopics / Methods

- Array with a number: `+ - * / **` applies to every element
- Array with array (same shape): works seat by seat
- Summary functions: `.sum()`, `.mean()`, `.max()`, `.min()`
- `axis` on 2D arrays: `axis=0` (down), `axis=1` (across)
- Leaving `axis` out means "use every number"

## Syntax

Rules:
- Two arrays must have the same shape, or one must be a single number.
- Methods need `()` to run. `shape` has none because it is a stored fact.
- `axis` goes inside the brackets: `t.sum(axis=0)`.
- The axis you name is the one that disappears.

### `axis` on a 2D array

````python
t = np.array([[1, 2, 3],
              [4, 5, 6]])
t.shape
````

````python
print(t.sum())          # everything
````

````python
t.sum(axis=0)    # down each column, one answer per column
````

````python
t.sum(axis=1)    # across each row, one answer per row
````

````python
t.sum(axis=0).shape, t.sum(axis=1).shape   # the named axis disappears
````

````python
t.max(axis=0)
````

````python
t.max(axis=1)
````

## Interview-style explanation

**Q: What does it mean that NumPy math is vectorized?**
One expression like `prices * 0.9` is applied to every element without writing a Python loop. It is shorter, and it runs faster because the looping happens in compiled code.

**Q: What does `a * b` do for two arrays?**
It multiplies element by element, so position 0 meets position 0, and so on. It is not a dot product or matrix multiplication. The shapes must match.

**Q: What does `axis` mean?**
It is the direction along which the function combines values. `axis=0` combines down the rows, which gives one result per column. `axis=1` combines across the columns, which gives one result per row. The named axis is the one that gets removed from the shape.

**Q: Shape is `(2, 3)`. What is the shape of `t.sum(axis=0)` and `t.sum(axis=1)`?**
`(3,)` and `(2,)`. Axis 0 removes the 2 and axis 1 removes the 3.

**Q: What happens if you leave `axis` out?**
The function uses all elements and returns one single number.

**Q: Why is `np.array([prices * 0.9])` wrong?**
`prices * 0.9` is already an array. Wrapping it in `[ ]` makes a list holding one array, and `np.array` turns that into a 2D array of shape `(1, 4)`. Use `np.array` only to convert a list.

## Coding examples

### Problem 1: Discount, 10% off then total
*Teaches: whole-array arithmetic followed by a summary function.*

````python
prices = np.array([200, 350, 120, 500])
discounted = prices * 0.9
print(discounted)
print(discounted.sum())
````

### Problem 2: Grace marks, add 5 then highest
*Teaches: the same pattern with `.max()`.*

````python
marks = np.array([40, 65, 80, 55, 90])
curved = marks + 5
print(curved.max())
````

### Problem 3: Sales, rows = shops, columns = days
*Teaches: choosing `axis` by asking which direction the numbers run in.*

````python
sales = np.array([[10, 20, 30],
                  [40, 50, 60]])
print(sales.sum(axis=0))   # total per day
print(sales.sum(axis=1))   # total per shop
````

### Problem 4: Marks, rows = students, columns = subjects
*Teaches: the same `axis` rule works for `.mean()` and `.max()`, and leaving `axis` out uses every number.*

````python
marks = np.array([[60, 70, 80],
                  [90, 50, 40],
                  [75, 85, 95]])
print(marks.mean(axis=1))   # average per student
print(marks.max(axis=0))    # highest per subject
print(marks.mean())         # overall average
````

### Problem 5: Bill, price * quantity then total
*Teaches: array-with-array math works seat by seat, then a summary function gives the total.*

````python
price = np.array([100, 200, 300])
qty   = np.array([2, 1, 4])
cost = price * qty
print(cost)
print(cost.sum())
````

### Problem 6: Celsius to Fahrenheit, then average per city
*Teaches: whole-array arithmetic on a 2D table, then `.mean()` with the right `axis`.*

````python
temps = np.array([[20, 30, 25],
                  [10, 15, 20]])
f = temps * 9 / 5 + 32
print(f)
print(f.mean(axis=1))
````

## Common mistakes

- Writing `.sum` without `()`: it prints the method instead of the total.
- Writing `.avg`: it does not exist, use `.mean()`.
- Swapping the axes: ask "which direction do the numbers I want to combine run in?"
- Writing `print(axis=1(a.sum()))`: `axis` goes inside, as in `a.sum(axis=1)`.
- Wrapping array math in `np.array([ ])` again: this adds an extra dimension.
- Using the original array after converting: use `f`, not `temps`.
