# NumPy Day 5: Useful Functions

**Topics:** `np.arange`, `sum`, `mean`, `max`, `min`, `np.sort`, `np.zeros`, `np.ones`, `axis=0` / `axis=1`

## Key Notes

- `np.arange(start, stop, step)` builds an array of numbers. Stop is **not** included (same logic as `range`). Spelled with one **r**.
- Counting down needs a **negative step**, and the stop must sit just past the last value you want.
- `np.zeros((rows, cols))` and `np.ones((rows, cols))`: the shape goes in its **own brackets**.
- `np.sort(arr)` returns a sorted **copy**; `arr.sort()` sorts the array **itself**.
- `axis=0` goes **down** (column totals), `axis=1` goes **sideways** (row totals).
- Array math must be saved with `=`, or the result is lost.

```python
import numpy as np
```

## Problem 1 (Easy)
Make an array of the odd numbers from 1 to 15. Print the array, then its sum.

```python
odds = np.arange(1, 16, 2)
print(odds)
print(odds.sum())
```

**Learned:** generate a number sequence with `np.arange` and save it in a variable to reuse it.

## Problem 2 (Easy)
Make an array of the multiples of 5 from 5 to 50. Print the array, then its mean.

```python
table5 = np.arange(5, 51, 5)
print(table5)
print(table5.mean())
```

**Learned:** use `np.arange` with a step, then call `.mean()` on the saved array.

## Problem 3 (Medium)
Prices are 10 to 100 in steps of 10. Rs 5 off every item. Print the new prices and their sum.

```python
prices = np.arange(10, 101, 10)
new_prices = prices - 5
print(new_prices)
print(new_prices.sum())
```

**Learned:** do math on a whole array in one line (`prices - 5`), and use the **new** array for the next step.

## Problem 4 (Medium)
Countdown from 50 to 5 in steps of 5. Print the array, then its max and min.

```python
countdown = np.arange(50, 4, -5)
print(countdown)
print(countdown.max())
print(countdown.min())
```

**Learned:** counting down needs a negative step, and the stop (4) sits just past the last value (5).

## Problem 5 (Hard)
Make 1 to 12, reshape into 3 rows and 4 columns. Print the table, the row totals, and the column totals.

```python
nums = np.arange(1, 13)
table = nums.reshape(3, 4)

print(table)
print(table.sum(axis=1))  # row totals
print(table.sum(axis=0))  # column totals
```

**Learned:** chain `arange`, `reshape` and `sum` with an axis to turn a flat list into a table and summarise it both ways.

## Problem 6 (Hard)
1. 3x4 table of zeros
2. 3x4 table of ones, then make every value 10
3. Sort `marks = [62, 45, 88, 71, 53]` smallest to largest

```python
grid = np.zeros((3, 4))
grid1 = np.ones((3, 4))
grid1 = grid1 + 9
marks = np.array([62, 45, 88, 71, 53])
marks.sort()

print(grid)
print(grid1)
print(marks)
```

**Learned:** ready-made arrays with `np.zeros` / `np.ones`, sorting with `.sort()`, and that array math must be saved with `=`.
