## What it is
When two arrays have different shapes, NumPy copies the smaller one to fill the bigger one. No loops needed.

## Rules to remember
1. **Row-style array** `[a, b, c]` (shape `(3,)` or `(1, 3)`): one number per column, copied **down** every row.
2. **Column-style array** `[[a], [b], [c]]` (shape `(3, 1)`): one number per row, copied **sideways** across every column.
3. Sizes must match, or be 1. Otherwise: `ValueError`.
4. Per-row average needs `keepdims=True` so the result stays a column.

**Memory trick:** a 1 means "copy me", and every other mismatch is an error.

## Shape reminder
- `np.array([2, 3, 4, 5])` has shape `(4,)` (1D, one pair of brackets)
- `np.array([[2, 3, 4, 5]])` has shape `(1, 4)` (2D, two pairs of brackets)

```python
import numpy as np
```

## Problem 1 (easy): add shipping per product
Prices table: rows are shops, columns are products. Add shipping of 10, 20, 30 to the three products in both shops.

**Status: Done**

```python
prices = np.array([[100, 200, 300],
                   [150, 250, 350]])
shipping = np.array([10, 20, 30])

print(prices + shipping)
```

**Learned:** a row-style array is copied down to every row.

## Problem 2 (easy): extra charge per shop
Shop 1 charges 10 extra and shop 2 charges 90 extra, on every product.

**Status: Done**

```python
prices = np.array([[100, 200, 300],
                   [150, 250, 350]])
extra = np.array([[10],
                  [90]])

print(prices + extra)
```

**Learned:** a column-style array `(2, 1)` is copied sideways across all columns.

## Problem 3 (medium): bonus per subject, moderation per student
Rows are students, columns are subjects (Maths, Science, English).
Bonus: +5 Maths, +10 Science, +0 English. Moderation: +2, +4, +6 for students 1, 2, 3.

**Status: Done**

```python
marks = np.array([[60, 70, 80],
                  [50, 90, 40],
                  [75, 65, 85]])

bonus = np.array([[5, 10, 0]])
moderation = np.array([[2], [4], [6]])

print(marks + bonus)
print(marks + moderation)
print(bonus + marks + moderation)
```

**Learned:** different directions need different shapes. Row-style for per-column, column-style for per-row. Both can be chained in one line.

## Problem 4 (hard): subtract each subject's average
Subtract the class average of each subject from every student's mark.

**Status: Done**

```python
marks = np.array([[60, 70, 80],
                  [50, 90, 40],
                  [75, 65, 85]])

avg = marks.mean(axis=0)
print(avg)
print(marks - avg)
```

**Learned:** a subject is a column, so average downward with `axis=0`. The result is row-style and broadcasts straight back.

## Problem 5 (hard): subtract each student's own average
For each student, subtract that student's average from all their marks.

**Status: Done**

```python
marks = np.array([[60, 70, 80],
                  [50, 90, 40],
                  [75, 65, 85]])

avg = marks.mean(axis=1, keepdims=True)
print(avg)
print(marks - avg)

# without keepdims the average is flat and copies downward (wrong direction)
print(marks.mean(axis=1))
```

**Learned:** a student is a row, so average sideways with `axis=1`. `keepdims=True` keeps it as a column so it copies sideways.

## Problem 6 (easy): delivery added to stock
A shop sells 3 items over 2 days (rows are days, columns are items). A delivery brings 5, 6, 7 extra of items 1, 2, 3. Add it to both days.

**Status: Done**

```python
sold = np.array([[10, 20, 30],
                 [40, 50, 60]])
delivery = np.array([5, 6, 7])

print(sold + delivery)
```

**Learned:** one number per column means a row-style array, copied down to every day.
