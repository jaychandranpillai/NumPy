# Day 6: NumPy Filtering (Boolean Masking)

**Core ideas**
- A comparison on an array gives a **mask**: an array of only True/False.
- `array[mask]` keeps only the items where the mask is True.
- True counts as 1 and False as 0, so `(mask).sum()` counts the True items.
- Combine conditions with `&` (and) / `|` (or). Each condition goes in its own `( )`. Do not use the words `and` / `or` on arrays.

**Patterns**

| Goal | Code |
|---|---|
| values that pass | `array[condition]` |
| how many pass | `(condition).sum()` |
| both true | `array[(c1) & (c2)]` |
| either true | `array[(c1) \| (c2)]` |
| filter, then math | `array[condition].mean()` / `.max()` |

## Mask vs filtered values

```python
import numpy as np

temps = np.array([28, 35, 31, 40, 25, 38])

print(temps > 30)           # mask: [False  True  True  True False  True]
print(temps[temps > 30])    # values: [35 31 40 38]
print((temps > 30).sum())   # count: 4
```

## Problem 1: temps above 30

**Teaches:** a comparison gives a mask, and the mask picks items.

```python
print(temps > 30)
print(temps[temps > 30])
print((temps > 30).sum())
```

## Problem 2: `>=` and `<`

**Teaches:** choosing the operator from the wording.

```python
marks = np.array([45, 82, 67, 91, 38, 74])
print(marks[marks >= 60])    # [82 67 91 74]
print((marks < 40).sum())    # 1
```

## Problem 3: `==` and `!=`

**Teaches:** equality works as a filter condition too.

```python
prices = np.array([120, 450, 80, 300, 450, 99])
print(prices[prices == 450])   # [450 450]
print(prices[prices != 450])   # [120  80 300  99]
print((prices == 450).sum())   # 2
```

## Problem 4: values template vs count template

**Teaches:** the wording decides which template to use.

```python
scores = np.array([55, 72, 90, 72, 40, 88])
print(scores[scores == 72])   # [72 72]
print(scores[scores > 70])    # [72 90 72 88]
print((scores > 70).sum())    # 4
```

## Problem 5: two conditions with `&` and `|`

**Teaches:** "between" = two comparisons joined with `&`, "either" = two joined with `|`.

```python
marks = np.array([45, 82, 67, 91, 38, 74, 60, 55])
print(marks[(marks >= 50) & (marks <= 80)])    # [67 74 60 55]
print(marks[(marks < 40) | (marks > 90)])      # [91 38] (original array order)
print(((marks >= 50) & (marks <= 80)).sum())   # 4
```

## Problem 6: ranges on ages

**Teaches:** "from X to Y" and "below X or above Y" as operators.

```python
ages = np.array([12, 25, 34, 17, 45, 60, 19, 30])
print(ages[(ages >= 18) & (ages <= 35)])     # [25 34 19 30]
print(ages[(ages < 18) | (ages > 50)])       # [12 17 60]
print(((ages >= 18) & (ages <= 35)).sum())   # 4
```

## Problem 7: filter first, then math

**Teaches:** do Day 4 math (`mean`, `max`) on only the items that passed.

```python
marks = np.array([45, 82, 67, 91, 38, 74, 60, 55, 29, 88])
print(marks[marks >= 50].mean())             # 73.857...
print(((marks < 40) | (marks > 80)).sum())   # 5
print(marks[marks < 80].max())               # 74
```

## Common slips
- Comparison alone gives a mask. Inside `array[ ]` it gives values.
- `.sum()` goes on the mask: `(cond).sum()`, not on the number.
- `[ ]` only right after the array name. Group conditions with `( )`.
- `==` means exactly one value. Use `>` `<` `>=` `<=` for above / below / from-to.
